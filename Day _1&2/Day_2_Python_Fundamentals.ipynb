```python
# A Simple Calculator to Understand Variables and if-else Statements

Num1 = float(input("Enter a Number"))
Num2 = float(input("Enter the Next Number"))

Operators =  ["+","-","*","/"]

operator = input("Enter the Operator :")

if operator in Operators:
    if operator == "+":
        print(Num1 + Num2)
    elif operator == "-":
        print(Num1 - Num2)
    elif operator == "*":
        print(Num1 * Num2)
    elif operator == "/":
        print(Num1 / Num2)

else:
    print("Please make sure whether the entered Operator is Correct!")
    
```

### Functions

```python
def calculate_average(nums):
    return sum(nums)/len(nums)

marks = [75,98,56,79,49,59]

print("Average :",calculate_average(marks))
```

```python
# Create a Function to identify prime number or not
def check_prime(n):
    if n<=1:
        return False

    for i in range(2,n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter a number : "))

if check_prime(num):
    print(num,"is a Prime Number")
else:
    print(num,"isn't a Prime Number")
    
```

```python
#Create a Function to Calculate Factorials
def factorial(n):

    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result


num = int(input("Enter a number: "))

if num < 0:
    print("Factorial is not defined for negative numbers")
else:
    print("Factorial:", factorial(num))
```

```python
#Create a Function to Find the Maximum Value in a List
def find_max(numbers):

    maximum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

    return maximum


numbers = [10, 50, 20, 80, 30]

print("Maximum value:", find_max(numbers))
```
