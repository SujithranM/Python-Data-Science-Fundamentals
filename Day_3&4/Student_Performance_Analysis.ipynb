```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    "Name":["John","David","Sarah","Alex","Kevin"],
    "Mathematics":[85,75,95,65,80],
    "Statistics":[90,82,92,70,85],
    "Programming":[80,88,96,75,90]   
    }
df = pd.DataFrame(data)
print(df)
```

```python
#Averages of the Students
df["Average"] = df[["Mathematics","Statistics","Programming"]].mean(axis=1)
df
```

```python
# Highest performing student

highest_Student=df.loc[df["Average"].idxmax()]
print("Highest Performing Student:")
print(highest_Student)
```

```python
# Average Marks for each subject
subject_averages = df[["Mathematics","Statistics","Programming"]].mean()
subject_averages
```

```python
#student scoring above 80 on average

above_80 = df[df["Average"]>80]

print(above_80[["Name","Average"]])
```

```python
# Bar chart comparing student performance

plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Average"])

plt.title("Student Performance Analysis")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.show()
```
